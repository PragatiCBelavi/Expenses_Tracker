```python
#Import Required Libraries
import datetime
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
#File created automatically
FILE_NAME = "expenses.csv"
```

```python
#Take user input and Save data to CSV file
def add_expense():
    date = datetime.date.today()
    category = input("Enter category: ")
    amount = float(input("Enter amount: "))

    with open(FILE_NAME, "a") as file:
        file.write(f"{date},{category},{amount}\n")

    print("Expense added!")
add_expense()
```

```python
#Read file and Show stored data
def view_expenses():
    with open(FILE_NAME, "r") as file:
        for line in file:
            print(line.strip())
view_expenses()
```

```python
#Calculate Total Expense
def total_expense():
    total = 0
    with open(FILE_NAME, "r") as file:
        for line in file:
            total += float(line.strip().split(",")[2])

    print("Total Spending:", total)
total_expense()
```

```python
#Prepare Data for Graph and You must group data category-wise.
def get_category_totals():
    data = {}
    with open(FILE_NAME, "r") as file:
        for line in file:
            _, category, amount = line.strip().split(",")
            amount = float(amount)
            data[category] = data.get(category, 0) + amount
    return data
```

```python
#expense trend over the month
def get_daily_totals(month, year):
    daily_totals = {}

    with open(FILE_NAME, "r") as file:
        for line in file:
            date, _, amount = line.strip().split(",")
            y, m, d = date.split("-")

            if y == year and m == month:
                daily_totals[int(d)] = daily_totals.get(int(d), 0) + float(amount)

    return daily_totals
```

```python
#line graph over monthly expenses
def show_monthly_line_graph():
    month = input("Enter month (MM): ")
    year = input("Enter year (YYYY): ")

    daily_data = get_daily_totals(month, year)

    if not daily_data:
        print("No data found for this month")
        return

    days = sorted(daily_data.keys())
    amounts = [daily_data[day] for day in days]

    plt.plot(days, amounts, marker='o')
    plt.xlabel("Day of Month")
    plt.ylabel("Amount Spent")
    plt.title(f"Daily Expenses for {month}-{year}")
    plt.grid(True)
    plt.show()
```

```python
#bar chart
def show_bar_graph():
    data = get_category_totals()
    plt.bar(data.keys(), data.values())
    plt.xlabel("Category")
    plt.ylabel("Amount")
    plt.title("Category-wise Expenses")
    plt.show()
show_bar_graph()
```

```python
#pie chart
def show_pie_chart():
    data = get_category_totals()
    plt.pie(data.values(), labels=data.keys(), autopct="%1.1f%%")
    plt.title("Expense Distribution")
    plt.show()
```

```python
##### Combine Everything Using Menu
def menu():
    while True:
        print("\n1. Add Expense")
        print("2. View Expenses")
        print("3. Total Expense")
        print("4. Bar Graph")
        print("5. Pie Chart")
        print("6. Monthly Line Graph")
        print("7. Exit")


        choice = input("Enter choice: ")

        if choice == "1":
            add_expense()
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            total_expense()
        elif choice == "4":
            show_bar_graph()
        elif choice == "5":
            show_pie_chart()
        elif choice == "6":
            show_monthly_line_graph()
        elif choice == "7":
            break
menu()
```
